### 1. Modify the color image imc so the rectangle is cyan instead of yellow. Which channel(s) need
to change?

```python
imc = np.zeros((50, 100, 3), dtype=np.uint8)
imc[:, :, 0] = 255
imc[10:40, 20:80, 1] = 255 #add green color to the image
plt.imshow(imc)
```

```python
# Change color
imc[10:40, 20:80, 0] = 0 #remove red color => Green
imc[10:40, 20:80, 2] = 255 #add blue color
plt.imshow(imc)
```

So, we need to change color chanel.

### 2. In the primaries image, slice out the bottom-right block (blue) using row and column ranges,and confirm with np.array_equal that it matches a fresh block filled with (0, 0, 255)

```python
imc2 = np.zeros((60, 90, 3), dtype=np.uint8)
# Create a white color
imc2[0:30, 30:60, :] = 255

# Create a red color
imc2[0:30, 60:, 0] = 255

# Create a yellow color
imc2[30:, 0:30, 0] = 255
imc2[30:, 0:30, 1] = 255

# Create a green color
imc2[30:, 30:60, 1] = 255

# Create a blue color
imc2[30:, 60:, 2] = 255
plt.imshow(imc2)
```
